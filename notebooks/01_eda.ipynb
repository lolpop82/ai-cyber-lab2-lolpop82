# Exploratory Data Analysis - Phishing URL Detection

This notebook explores the phishing URL feature dataset to understand class distributions, feature characteristics, and relationships between features.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data import load_data

sns.set_theme(style='whitegrid')
df = load_data()
print(f'Dataset shape: {df.shape}')
df.head()
```

## Class Distribution

```python
fig, ax = plt.subplots(figsize=(6, 4))
counts = df['label'].value_counts()
counts.index = ['Benign (0)', 'Phishing (1)']
counts.plot(kind='bar', color=['steelblue', 'salmon'], ax=ax)
ax.set_title('Class Distribution')
ax.set_xlabel('Class')
ax.set_ylabel('Count')
for i, v in enumerate(counts):
    ax.text(i, v + 10, str(v), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

## Feature Summaries

```python
df.describe().round(2)
```

```python
feature_cols = [c for c in df.columns if c != 'label']

fig, axes = plt.subplots(2, 5, figsize=(18, 8))
for ax, col in zip(axes.flatten(), feature_cols):
    for label, color in [(0, 'steelblue'), (1, 'salmon')]:
        subset = df[df['label'] == label]
        ax.hist(subset[col], bins=25, alpha=0.6, color=color,
                label='Benign' if label == 0 else 'Phishing')
    ax.set_title(col)
    ax.legend(fontsize=7)
plt.suptitle('Feature Distributions by Class', fontsize=14)
plt.tight_layout()
plt.show()
```

## Insight

The dataset is balanced with an equal number of phishing and benign samples (1000 each). Phishing URLs tend to be significantly longer (mean ~80 chars vs ~45 for benign) and contain more special characters and subdomains. Notably, phishing URLs are far less likely to use HTTPS (~35%) compared to benign URLs (~85%), and they are much more likely to contain an IP address in the URL or suspicious keywords like 'login' and 'verify'. These feature separations suggest that even a simple baseline classifier should achieve reasonable performance on this task.
